```python
# cpu, cuda, mps (Apple Silicon)

if torch.cuda.is_available():
    device = torch.device("cuda")
else:
    device = torch.device("cpu")

device = torch.device("mps") # für Apple Silicon 
device
```

```python
import torch.nn as nn # Neuronale Netze
import torch.optim as optim # Optimierer
# 10 Eingänge, 1 Ausgang -> Fully-connected feed-forward neural network
NN = nn.Linear(10,1) # im Tensorflow -> Dense 
# MLP, P -> Multi-layer Perceptron (old-school)
# in Publikationen: FC, FFN, FFNN 
loss = nn.MSELoss() # Kostenfunktion, Verlustfunktion
# loss, cost, criterion, crit
optimizer = optim.SGD(NN.parameters(), lr = 1e-2 ) # Stochastic Gradient Descent
# lr = Lernrate (eta, alpha); NN.parameters() -> Gewichte, Parameter von Modell
input_data = torch.randn(10) # zufällige Daten (X)
output = NN(input_data) # y_pred, y_hat -> Modell 
y = torch.ones(1) # target, bzw. wir möchten, dass NN eine 1 ausgibt -> Realität, y_true
loss_output = loss(output, y) # Unterschied zwischen Modell und Realität , Anfangswert der Kostenfunktion

print(f"Kosten: {loss_output:.2} (vor Update)")

# zwei magische Zeilen (Gradientenabstiegsmethode)
loss_output.backward() # Gradienten berechnen
optimizer.step() # Update-Schritt 
output_new = NN(input_data) # bereits mit neuen Gewichten 
loss_new = loss(output_new,y) # Wert der Kostenfunktion nach Update
print(f"Kosten: {loss_new:.2} (nach Update)")
```
